```python

for id in llm['patent_id']:
    if id == "9999846":
        continue
    if id[-2] == '_':
        for name in name_groups:
            llm.loc[llm['patent_id']==id,f"our_score_{name}"] = our[our['Num_patent']==int(id[:-2])][f"Similarity_{name}"].iloc[0]
    else:
        #print(id)
        for name in name_groups:
            llm.loc[llm['patent_id']==id,f"our_score_{name}"] = our[our['Num_patent']==int(id)][f"Similarity_{name}"].iloc[0]


for id in llm['patent_id']:
    if id == "9999846":
        continue
    if id[-2] == '_':
        for name in name_groups:
            llm.loc[llm['patent_id']==id,f"their_score_{name}"] = ai_model_predictions[ai_model_predictions['Num_patent']==int(id[:-2])][f"ai_score_{name}"].iloc[0]
    else:
        #print(id)
        for name in name_groups:
            llm.loc[llm['patent_id']==id,f"their_score_{name}"] = ai_model_predictions[ai_model_predictions['Num_patent']==int(id)][f"ai_score_{name}"].iloc[0]
```

```python
from briertools import BrierScorer

group = ['ml','vision','nlp','speech','evo','hardware','kr','planning']

for name in group:
    y_true = []
    for row in llm.subdomain_scores:
        if list(ast.literal_eval(row).values())[group.index(name)] == None:
            y_true.append(0)
        else:
            y_true.append(list(ast.literal_eval(row).values())[group.index(name)])
    y_pred_our = list(llm[f"our_score_{name}"])
    y_pred_their = list(llm[f"their_score_{name}"])

    brier_scorer = BrierScorer()
    brier_score_our = brier_scorer.score(y_true, y_pred_our)
    brier_score_their = brier_scorer.score(y_true, y_pred_their)

    print('-------------')
    print(f"group: {name}")
    print(f"Our brierscore: {brier_score_our}")
    print(f"Their brierscore: {brier_score_their}")

[0.0999191699455435,0.03480931648603065,0.027610142622806873,0.015259631669048215,0.009260906791685761,0.04189305516983291,0.03270597378330232,0.02941113938968637]
[0.12077165485348146,0.06686572267865062,0.0819477500247959,0.051533949947147294,0.014714549640579415,0.24272630178268487,0.099271725405758,0.1385584520508763]    
```

# Plotes

```python
import matplotlib.pyplot as plt
import numpy as np

  
# Data
groups = ['ML', 'Vision', 'NLP', 'Speech', 'EVO', 'Hardware', 'KR', 'Planning']
our_scores = [0.0999191699455435,0.03480931648603065,0.027610142622806873,0.015259631669048215,0.009260906791685761,0.04189305516983291,0.03270597378330232,0.02941113938968637]
    
their_scores = [0.12077165485348146,0.06686572267865062,0.0819477500247959,0.051533949947147294,0.014714549640579415,0.24272630178268487,0.099271725405758,0.1385584520508763]

# Setup
x = np.arange(len(groups))
width = 0.35

fig, ax = plt.subplots(figsize=(12, 6))

# Bars
bars1 = ax.bar(x - width/2, our_scores, width, label='UIS Pipeline', 
               color='#2ca02c', edgecolor='white', linewidth=1.2)
bars2 = ax.bar(x + width/2, their_scores, width, label='AIPD 2023', 
               color='#d62728', edgecolor='white', linewidth=1.2)

# Labels & title
ax.set_ylabel('Brier Score (lower is better)', fontsize=14, fontweight='bold')
ax.set_xlabel('Group', fontsize=14, fontweight='bold')
ax.set_title('Model Performance Comparison Across Domains', fontsize=16, fontweight='bold', pad=20)
ax.set_xticks(x)
ax.set_xticklabels(groups, fontsize=12, fontweight='bold')
ax.legend(loc='upper right', frameon=True, fancybox=True, shadow=True)

# Add value labels on bars
for bar in bars1 + bars2:
    height = bar.get_height()
    ax.annotate(f'{height:.3f}',
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3),
                textcoords="offset points",
                ha='center', va='bottom', fontsize=11, fontweight='bold', rotation=45)

# Grid & styling
ax.grid(axis='y', alpha=0.3, linestyle='--')
ax.set_axisbelow(True)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Improvement annotations (optional)
for i, (our, their) in enumerate(zip(our_scores, their_scores)):
    improvement = ((their - our) / their) * 100
    if improvement > 0:
        if i == 4:
            ax.annotate(f'↓ {improvement:.0f}%', 
                        xy=(x[i], max(our, their) + 0.025),
                        ha='right', fontsize=10, color='green', fontweight='bold')
        else:
            ax.annotate(f'↓ {improvement:.0f}%', 
                        xy=(x[i], max(our, their) + 0.01),
                        ha='right', fontsize=10, color='green', fontweight='bold')

plt.tight_layout()
plt.show()

print(f"Mean Our Barier score: {np.mean(our_scores)}")
print(f"Mean Their Barier score: {np.mean(their_scores)}")
# Optional: Save figure
fig.savefig('brier_scores_comparison.png', dpi=300, bbox_inches='tight')
```

```python
import ast

list_dist = []
for i in range(len(llm)):
    # Get the subdomain_scores and replace None with 0
    subdomain_scores = ast.literal_eval(llm.iloc[i]['subdomain_scores'])
    
    # Replace None values with 0
    subdomain_scores = {k: (v if v is not None else 0) for k, v in subdomain_scores.items()}
    
    a = max(subdomain_scores.values())
    b = llm['overall_ai_score'].iloc[i]
    list_dist.append(b - a)
```

```python
list_dist
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(10, 6))
plt.hist(list_dist, bins=40, edgecolor='black', alpha=0.7)

# Add summary statistics
mean_val = np.mean(list_dist)
median_val = np.median(list_dist)
std_val = np.std(list_dist)

plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.3f}')
plt.axvline(median_val, color='green', linestyle='--', linewidth=2, label=f'Median: {median_val:.3f}')

plt.xlabel('Score Difference')
plt.ylabel('Frequency')
plt.title(f'Distribution of Score Differences\n(Std: {std_val:.3f})')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Print statistics
print(f"Mean: {mean_val:.4f}")
print(f"Median: {median_val:.4f}")
print(f"Std: {std_val:.4f}")
print(f"Min: {np.min(list_dist):.4f}")
print(f"Max: {np.max(list_dist):.4f}")
```

```python
mean_abs = np.mean(np.abs(list_dist))
print(f"Mean of absolute values: {mean_abs:.4f}")
```
